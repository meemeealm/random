# Startup User Acquisition  


You are analyzing the growth trajectories of two different startups over a period of 10 days ($t = 0$ to $t = 9$):

Startup Alpha (Linear Growth): Starts with 100 users and acquires a steady 50 new users every day.

Startup Beta (Exponential Growth): Starts with 100 users and grows at a rate of 30% ($0.30$) compounded daily.

### Tasks:  
1. Generate an array of days from $0$ to $9$ using numpy.arange.  

2. Calculate the number of users for Startup Alpha each day using the linear formula:$Users_{linear} = Start + (Rate \times t)$  

3. Calculate the number of users for Startup Beta each day using the exponential formula:$Users_{exponential} = Start \times (1 + GrowthRate)^t$  
4. Find the exact day when Startup Beta's user count surpasses Startup Alpha's user count.

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
days = np.arange(10)

initial_users = 100
```

```python
# calculate linear growth for startup Alpha
daily_addition = 50
users_alpha = initial_users + (daily_addition + days)
```

```python
# calculate growth for startup Beta
growth_rate = 0.3
exponential_users = initial_users * ((1+growth_rate)**days)
```

```python
print("Day | Startup Alpha (Linear) | Startup Beta (Exponential)")
print("-" * 50)
for d, lin, exp in zip(days, users_alpha, exponential_users):
    print(f" {d}  |          {lin:5.0f}         |          {exp:6.1f}")
```

```python
surpass_day = np.where(exponential_users > users_alpha)[0][0]
surpass_users = exponential_users[surpass_day]
```

```python
plt.figure(figsize=(10, 6))

plt.plot(days, users_alpha, label='Startup Alpha (Linear: +50/day)', color='blue', marker='o', linewidth=2)
plt.plot(days, exponential_users, label='Startup Beta (Exponential: +30%/day)', color='red', marker='s', linewidth=2)

# Highlight the crossover point
plt.axvline(x=surpass_day, color='green', linestyle='--', alpha=0.7, label=f'Crossover on Day {surpass_day}')
plt.scatter(surpass_day, surpass_users, color='green', zorder=5, s=100)

# Labels and Styling
plt.title('Startup User Acquisition: Linear vs. Exponential Growth', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Days', fontsize=12)
plt.ylabel('Total Users', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)

plt.tight_layout()
plt.show()
```
